```python
import pandas as pd
df = pd.read_excel("Goal1.xlsx", sheet_name="data")
print("Dataset loaded successfully!")
display(df.head())
```

```python
print("Year range:")
print(df["TimePeriod"].min(), "-", df["TimePeriod"].max())
print("\nNumber of geographical areas:")
print(df["GeoAreaName"].nunique())
print("\nSample countries/regions:")
print(df["GeoAreaName"].unique()[:20])
```

```python
import matplotlib.pyplot as plt
india = df[df["GeoAreaName"] == "India"].copy()
print("Available indicators for India:")
print(india["SeriesDescription"].unique())
```

```python
indicator = india["SeriesDescription"].dropna().iloc[0]
trend = india[india["SeriesDescription"] == indicator].copy()
trend = trend.sort_values("TimePeriod")
plt.figure(figsize=(10, 5))
plt.plot(trend["TimePeriod"], trend["Value"], marker="o")
plt.title(f"India: {indicator}")
plt.xlabel("Year")
plt.ylabel("Reported Value")
plt.grid(True)
plt.tight_layout()
plt.show()
print("Indicator:", indicator)
```

```python
# Check available indicators and their units
display(
    df[["SeriesDescription", "Units"]]
    .drop_duplicates()
    .reset_index(drop=True)
)
```

```python
indicator_summary = (
    df.groupby(["SeriesDescription", "Units"])
    .agg(
        records=("Value", "count"),
        countries=("GeoAreaName", "nunique"),
        years=("TimePeriod", "nunique")
    )
    .reset_index()
    .sort_values("records", ascending=False)
)

display(indicator_summary)
```

```python
# Show available indicators for India
india = df[df["GeoAreaName"].astype(str).str.strip() == "India"].copy()

display(
    india[["SeriesDescription", "Units", "TimePeriod", "Value"]]
    .sort_values(["SeriesDescription", "TimePeriod"])
    .head(20)
)
```

```python
# Remove rows without year or target value
india = india.dropna(subset=["TimePeriod", "Value"])
# Sort observations chronologically
india = india.sort_values("TimePeriod")
print("Total usable observations:", len(india))
print("Years available:", sorted(india["TimePeriod"].unique()))
```

```python
print("Observations per indicator:")
print(
    india.groupby("SeriesDescription")["Value"]
    .count()
    .sort_values(ascending=False)
)
```

```python
indicator_counts = (
    india.groupby("SeriesDescription")
    .agg(
        records=("Value", "count"),
        years=("TimePeriod", "nunique")
    )
    .reset_index()
    .sort_values("records", ascending=False)
)
display(indicator_counts.head(10))
```

```python
import matplotlib.pyplot as plt
# Select the indicator with the most records
best_indicator = indicator_counts.iloc[0]["SeriesDescription"]
plot_data = india[
    india["SeriesDescription"] == best_indicator
].copy()
plot_data = plot_data.sort_values("TimePeriod")
plt.figure(figsize=(10, 5))
plt.plot(
    plot_data["TimePeriod"],
    plot_data["Value"],
    marker="o"
)
plt.title(f"India: {best_indicator}")
plt.xlabel("Year")
plt.ylabel("Value")
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Select required columns
model_data = plot_data[["TimePeriod", "Value"]].copy()
# Convert columns to numeric
model_data["TimePeriod"] = pd.to_numeric(
    model_data["TimePeriod"], errors="coerce"
)
model_data["Value"] = pd.to_numeric(
    model_data["Value"], errors="coerce"
)
# Remove missing values
model_data = model_data.dropna()
# Group multiple observations from the same year
model_data = (
    model_data.groupby("TimePeriod", as_index=False)["Value"]
    .mean()
    .sort_values("TimePeriod")
)
print("Prepared data:")
display(model_data)
print("Total yearly records:", len(model_data))
```

```python
from sklearn.model_selection import train_test_split
# Input feature and target
X = model_data[["TimePeriod"]]
y = model_data["Value"]
# Keep chronological order for time-series data
split_index = int(len(model_data) * 0.8)
X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]
y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]
print("Training records:", len(X_train))
print("Testing records:", len(X_test))
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score

# Create and train model
model = LinearRegression()
model.fit(X_train, y_train)

# Predict test values
y_pred = model.predict(X_test)

# Evaluate model
mae = mean_absolute_error(y_test, y_pred)

print("Mean Absolute Error:", round(mae, 3))

if len(y_test) >= 2:
    print("R2 Score:", round(r2_score(y_test, y_pred), 3))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.plot(
    X_test["TimePeriod"],
    y_test,
    marker="o",
    label="Actual Values"
)

plt.plot(
    X_test["TimePeriod"],
    y_pred,
    marker="x",
    linestyle="--",
    label="Predicted Values"
)

plt.title("India SDG Goal 1: Actual vs Predicted")
plt.xlabel("Year")
plt.ylabel("Indicator Value")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
# Find the latest year in your data
last_year = int(model_data["TimePeriod"].max())
# Create the next 5 years
future_years = pd.DataFrame({
    "TimePeriod": np.arange(last_year + 1, last_year + 6)
})
# Predict future values
future_predictions = model.predict(future_years)
# Display results
future_results = pd.DataFrame({
    "Year": future_years["TimePeriod"],
    "Predicted_Value": future_predictions
})
display(future_results)
```

```python
plt.figure(figsize=(10, 5))
plt.plot(
    model_data["TimePeriod"],
    model_data["Value"],
    marker="o",
    label="Historical Data"
)
plt.plot(
    future_results["Year"],
    future_results["Predicted_Value"],
    marker="x",
    linestyle="--",
    label="Future Predictions"
)
plt.title("SDG Goal 1: Future Indicator Prediction")
plt.xlabel("Year")
plt.ylabel("Indicator Value")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import mean_absolute_error
import numpy as np

# Baseline: predict the last training value for every test year
baseline_pred = np.full(
    len(y_test),
    y_train.iloc[-1]
)

model_mae = mean_absolute_error(y_test, y_pred)
baseline_mae = mean_absolute_error(y_test, baseline_pred)

print("Linear Regression MAE:", round(model_mae, 3))
print("Baseline MAE:", round(baseline_mae, 3))

if model_mae < baseline_mae:
    print("Linear Regression performs better than baseline.")
else:
    print("Linear Regression does not outperform the baseline.")
```
